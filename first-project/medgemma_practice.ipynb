## Step 1: get the data

We download the first batch of **ChestX-ray14**, a free dataset from the U.S. National Institutes of Health (NIH): about 5,000 chest X-rays (around 2 GB), plus a label file saying what doctors' reports found in each one.

Source: https://nihcc.app.box.com/v/ChestXray-NIHCC. This takes a few minutes.

```python
import os, tarfile, urllib.request
from pathlib import Path

DATA = Path("data")
DATA.mkdir(exist_ok=True)

# Direct download links for the NIH files (the same links the open-source PyHealth library uses)
LABELS_URL = ("https://nihcc.app.box.com/index.php?rm=box_download_shared_file"
              "&vanity_name=ChestXray-NIHCC&file_id=f_219760887468")
IMAGES_URL = "https://nihcc.box.com/shared/static/vfk49d74nhbxq3nqjg0900w5nvkorp5c.gz"  # images_001

labels_file = DATA / "Data_Entry_2017_v2020.csv"
if not labels_file.exists():
    print("Downloading the label file ...")
    urllib.request.urlretrieve(LABELS_URL, labels_file)

images_dir = DATA / "images"
if not images_dir.exists():
    tar_path = DATA / "images_001.tar.gz"
    print("Downloading about 2 GB of X-rays (a few minutes) ...")
    urllib.request.urlretrieve(IMAGES_URL, tar_path)
    print("Unpacking ...")
    with tarfile.open(tar_path) as tar:
        try:
            tar.extractall(DATA, filter="data")   # safe extraction (newer Python)
        except TypeError:
            tar.extractall(DATA)                  # older Python without the filter option
    tar_path.unlink()  # free up disk space

print("X-rays downloaded:", len(list(images_dir.glob("*.png"))))
```

## Step 2: build a small test set

We pick **10 X-rays labeled "Effusion"** (fluid around the lungs) and **10 labeled "No Finding"** (healthy). A fixed random seed means everyone running this notebook gets the same 20 images, so you can compare results with classmates.

Then we look at them. Can *you* spot the fluid?

```python
import csv, random
import matplotlib.pyplot as plt
from PIL import Image

FINDING = "Effusion"   # try "Cardiomegaly" or "Pneumothorax" later
N = 10                 # images per group

downloaded = {p.name for p in images_dir.glob("*.png")}
yes_imgs, no_imgs = [], []
with open(labels_file, newline="") as f:
    for row in csv.DictReader(f):
        if row["Image Index"] not in downloaded:
            continue
        found = row["Finding Labels"].split("|")
        if FINDING in found:
            yes_imgs.append(row["Image Index"])
        elif found == ["No Finding"]:
            no_imgs.append(row["Image Index"])

print(f"Available: {len(yes_imgs)} with {FINDING}, {len(no_imgs)} with No Finding")

random.seed(42)
test_set = [(name, "yes") for name in random.sample(yes_imgs, N)] + \
           [(name, "no") for name in random.sample(no_imgs, N)]
random.shuffle(test_set)

fig, axes = plt.subplots(4, 5, figsize=(15, 12))
for ax, (name, truth) in zip(axes.flat, test_set):
    ax.imshow(Image.open(images_dir / name), cmap="gray")
    ax.set_title(f"{name}\nreal answer: {truth}", fontsize=9)
    ax.axis("off")
plt.tight_layout()
plt.show()
```

## Step 3: ask the model

We load **MedGemma 4B**, Google's medical vision-language model, and ask it the same yes/no question about each X-ray. Loading takes a few minutes the first time. Then each image takes a few seconds.

The wording of the question matters. Changing it is step 2 of the "try first, train later" ladder: better instructions.

```python
from transformers import pipeline

# bfloat16 is best; the free T4 GPU doesn't support it, so we fall back to float16 there.
dtype = torch.bfloat16 if torch.cuda.is_available() and torch.cuda.is_bf16_supported() else torch.float16
pipe = pipeline("image-text-to-text", model="google/medgemma-4b-it", torch_dtype=dtype, device="cuda")

QUESTION = ("Is there a pleural effusion (fluid around the lungs) in this chest X-ray? "
            "Answer with only one word: yes or no.")

def to_yes_no(text):
    words = text.strip().lower().split()
    first = words[0].strip(".,!:;*\"'") if words else ""
    return first if first in ("yes", "no") else "unclear"

results = []
for i, (name, truth) in enumerate(test_set, start=1):
    image = Image.open(images_dir / name).convert("RGB")
    messages = [
        {"role": "system", "content": [{"type": "text", "text": "You are an expert radiologist."}]},
        {"role": "user", "content": [{"type": "text", "text": QUESTION}, {"type": "image", "image": image}]},
    ]
    reply = pipe(text=messages, max_new_tokens=20, do_sample=False)[0]["generated_text"][-1]["content"].strip()  # same answer every run
    answer = to_yes_no(reply)
    results.append({"image": name, "truth": truth, "answer": answer, "reply": reply})
    mark = "right" if answer == truth else "WRONG"
    print(f"[{i:2}/20] {name}  real: {truth:3}  model: {answer:7}  {mark}   (said: {reply!r})")
```

## Step 4: score it

In medicine, the *kind* of mistake matters:

- **Missed it:** the fluid was there, but the model said no. A sick patient could be sent home.
- **False alarm:** the lungs were healthy, but the model said yes. A patient could get worried or have extra tests.

```python
caught      = sum(r["truth"] == "yes" and r["answer"] == "yes" for r in results)
missed      = sum(r["truth"] == "yes" and r["answer"] != "yes" for r in results)
false_alarm = sum(r["truth"] == "no"  and r["answer"] == "yes" for r in results)
cleared     = sum(r["truth"] == "no"  and r["answer"] == "no"  for r in results)
unclear     = sum(r["answer"] == "unclear" for r in results)

print(f"{'':14}{'model: yes':>12}{'model: no/unclear':>20}")
print(f"{'really yes':14}{caught:>12}{missed:>20}    caught / MISSED")
print(f"{'really no':14}{false_alarm:>12}{N - false_alarm:>20}    FALSE ALARM / cleared")
print()
print(f"Accuracy:    {(caught + cleared) / len(results):.0%}   ({caught + cleared} of {len(results)} right)")
print(f"Sensitivity: {caught / N:.0%}   (real cases caught: {caught} of {N})")
print(f"Specificity: {cleared / N:.0%}   (healthy lungs cleared: {cleared} of {N})")
if unclear:
    print(f"Unclear:     {unclear} answers weren't a clear yes or no; they count as wrong.")

# Look at the mistakes: seeing HOW a model is wrong teaches you the most
wrong = [r for r in results if r["answer"] != r["truth"]]
if wrong:
    fig, axes = plt.subplots(1, len(wrong), figsize=(3 * len(wrong), 3.5), squeeze=False)
    for ax, r in zip(axes[0], wrong):
        ax.imshow(Image.open(images_dir / r["image"]), cmap="gray")
        kind = "MISSED" if r["truth"] == "yes" else "FALSE ALARM"
        ax.set_title(f"{kind}\nmodel said: {r['reply'][:20]}", fontsize=9)
        ax.axis("off")
    plt.tight_layout()
    plt.show()
```

## Step 5 (optional): try different prompts

The words you use to ask the question can change the score a lot. This step compares **four prompt styles**:

| Style | What we add to the question |
|---|---|
| 1. Plain yes/no | Nothing. This is the baseline from Step 3. |
| 2. Give it a definition | What fluid around the lungs looks like on an X-ray |
| 3. Describe, then decide | "First describe what you see, then answer." You can read *why* it answered. |
| 4. Confidence number | "How likely is it, from 0 to 100?" This lets us compute **AUC**. |

**Two rules make this a fair test:**

1. **More images.** With 20 images, one image moves the score by 5 points, so a "winning" prompt could just be luck. Here we use 25 + 25 images to compare prompts, and a *different* 25 + 25 to check the winner.
2. **Don't grade on the practice test.** If you keep rewording until the score goes up on the same images, you've only fit the prompt to those images. So we pick the best prompt on the **tuning set**, then confirm it on a **fresh set** it has never seen. We also keep each patient's X-rays in only one of the two sets.

Every answer is generated the same way each run (no randomness), so differences come from the prompt, not from luck. **This takes about 20 to 40 minutes on the free T4 GPU.** You can lower `PER_GROUP` to go faster.

```python
import re

PER_GROUP = 25   # images per group (fluid / healthy) in EACH of the two sets

# --- Pick new images, keeping every patient in only one set ---
already_used = {name for name, _ in test_set}
patient = lambda name: name.split("_")[0]      # "00000013_005.png" -> patient "00000013"

random.seed(7)
yes_pool = [n for n in yes_imgs if n not in already_used]
no_pool  = [n for n in no_imgs  if n not in already_used]
patients = sorted({patient(n) for n in yes_pool + no_pool})
random.shuffle(patients)
tune_patients = set(patients[: len(patients) // 2])

def pick(pool, in_tune, k):
    names = [n for n in pool if (patient(n) in tune_patients) == in_tune]
    return random.sample(names, k)

tune_set    = [(n, "yes") for n in pick(yes_pool, True, PER_GROUP)]  + [(n, "no") for n in pick(no_pool, True, PER_GROUP)]
confirm_set = [(n, "yes") for n in pick(yes_pool, False, PER_GROUP)] + [(n, "no") for n in pick(no_pool, False, PER_GROUP)]
print(f"Tuning set: {len(tune_set)} images.  Confirm set: {len(confirm_set)} images.  No patient appears in both.")
```

```python
# --- The four prompt styles. Each one has its own way of reading the model's reply. ---
FLUID = "a pleural effusion (fluid around the lungs)"

def read_yes_no(reply):
    return to_yes_no(reply), None

def read_final_answer(reply):
    found = re.findall(r"final answer\W*(yes|no)", reply.lower())
    return (found[-1] if found else "unclear"), None

def read_confidence(reply):
    found = re.search(r"\d{1,3}", reply)
    if not found:
        return "unclear", None
    score = min(int(found.group()), 100)
    return ("yes" if score >= 50 else "no"), score

PROMPTS = {
    "1. Plain yes/no": (
        f"Is there {FLUID} in this chest X-ray? Answer with only one word: yes or no.",
        read_yes_no, 20),
    "2. Give it a definition": (
        f"A pleural effusion is fluid between the lung and the chest wall. On a frontal chest X-ray it "
        f"usually shows as blunting of the angle where the lung meets the diaphragm, or a hazy layer at "
        f"the base of the lung that curves up the side. Is there {FLUID} in this chest X-ray? "
        f"Answer with only one word: yes or no.",
        read_yes_no, 20),
    "3. Describe, then decide": (
        f"First, describe what you see at the bases of both lungs and at the angles where the lungs meet "
        f"the diaphragm, in two or three sentences. Then, on a new line, write 'Final answer: yes' or "
        f"'Final answer: no' to say whether there is {FLUID}.",
        read_final_answer, 200),
    "4. Confidence number": (
        f"How likely is it that this chest X-ray shows {FLUID}? Reply with only a number from 0 "
        f"(definitely not) to 100 (definitely yes).",
        read_confidence, 10),
}

def ask(image_name, prompt, max_tokens):
    image = Image.open(images_dir / image_name).convert("RGB")
    messages = [
        {"role": "system", "content": [{"type": "text", "text": "You are an expert radiologist."}]},
        {"role": "user", "content": [{"type": "text", "text": prompt}, {"type": "image", "image": image}]},
    ]
    out = pipe(text=messages, max_new_tokens=max_tokens, do_sample=False)   # do_sample=False: same answer every run
    return out[0]["generated_text"][-1]["content"].strip()

def auc(rows):
    # Chance that a random real case gets a higher number than a random healthy one (ties count half)
    pos = [r["score"] for r in rows if r["truth"] == "yes" and r["score"] is not None]
    neg = [r["score"] for r in rows if r["truth"] == "no" and r["score"] is not None]
    if not pos or not neg:
        return None
    wins = sum((p > n) + 0.5 * (p == n) for p in pos for n in neg)
    return wins / (len(pos) * len(neg))

def score(rows):
    yes_n = sum(r["truth"] == "yes" for r in rows)
    no_n  = sum(r["truth"] == "no" for r in rows)
    caught  = sum(r["truth"] == "yes" and r["answer"] == "yes" for r in rows)
    cleared = sum(r["truth"] == "no"  and r["answer"] == "no"  for r in rows)
    return {"accuracy": (caught + cleared) / len(rows), "sensitivity": caught / yes_n,
            "specificity": cleared / no_n, "balanced": (caught / yes_n + cleared / no_n) / 2,
            "unclear": sum(r["answer"] == "unclear" for r in rows), "auc": auc(rows)}

def run_prompt(style, image_set):
    prompt, reader, max_tokens = PROMPTS[style]
    rows = []
    for name, truth in image_set:
        reply = ask(name, prompt, max_tokens)
        answer, number = reader(reply)
        rows.append({"image": name, "truth": truth, "answer": answer, "score": number, "reply": reply})
    return rows

print("Ready:", ", ".join(PROMPTS))
```

```python
# --- Compare all four prompts on the TUNING set (this is the slow part) ---
tune_results, tune_scores = {}, {}
for style in PROMPTS:
    print(f"Running {style} on {len(tune_set)} images ...")
    tune_results[style] = run_prompt(style, tune_set)
    tune_scores[style] = score(tune_results[style])

print(f"\n{'Prompt style':28}{'Accuracy':>10}{'Sensitivity':>13}{'Specificity':>13}{'AUC':>7}{'Unclear':>9}")
for style, s in tune_scores.items():
    auc_text = f"{s['auc']:.2f}" if s["auc"] is not None else "  -"
    print(f"{style:28}{s['accuracy']:>10.0%}{s['sensitivity']:>13.0%}{s['specificity']:>13.0%}{auc_text:>7}{s['unclear']:>9}")

# Bar chart: sensitivity (missed cases) and specificity (false alarms) side by side
styles = list(tune_scores)
xs = range(len(styles))
fig, ax = plt.subplots(figsize=(10, 4))
ax.bar([x - 0.2 for x in xs], [tune_scores[s]["sensitivity"] for s in styles], width=0.4, label="Sensitivity (catches real cases)")
ax.bar([x + 0.2 for x in xs], [tune_scores[s]["specificity"] for s in styles], width=0.4, label="Specificity (avoids false alarms)")
ax.set_xticks(list(xs))
ax.set_xticklabels(styles)
ax.set_ylim(0, 1)
ax.set_ylabel("share of images")
ax.set_title("Same model, same images, different wording")
ax.legend()
plt.tight_layout()
plt.show()

# Read a few "describe, then decide" answers: you can see the model's reasoning
for r in tune_results["3. Describe, then decide"][:3]:
    print(f"\n{r['image']} (real answer: {r['truth']}):\n{r['reply']}")
```

```python
# --- Pick the winner on the tuning set, then CHECK it on the fresh confirm set ---
# "Balanced" = the average of sensitivity and specificity, so a prompt can't win by always saying "no".
best = max(tune_scores, key=lambda s: tune_scores[s]["balanced"])
print(f"Best on the tuning set: {best}  (balanced score {tune_scores[best]['balanced']:.0%})")

confirm_rows = run_prompt(best, confirm_set)
baseline_rows = run_prompt("1. Plain yes/no", confirm_set)
for label, rows in [(best + "  (winner)", confirm_rows), ("1. Plain yes/no  (baseline)", baseline_rows)]:
    s = score(rows)
    print(f"Confirm set, {label:40} accuracy {s['accuracy']:.0%}, sensitivity {s['sensitivity']:.0%}, "
          f"specificity {s['specificity']:.0%}, balanced {s['balanced']:.0%}")

print("\nIf the winner still beats the baseline on images it has never seen, the better prompt is probably real.")
print("If the gap shrinks or flips, the 'win' on the tuning set was mostly luck. That is a real finding too.")
```
